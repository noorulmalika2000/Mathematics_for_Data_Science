## Task 1 – Linear Algebra Implementation

In this task, we use NumPy to perform mathematical operations on vectors and matrices.

```python
# Define two vectors
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
print("Vector A:", vector_a)
print("Vector B:", vector_b)
# Vector addition
print("\nVector Addition:", vector_a + vector_b)
# Vector subtraction
print("Vector Subtraction:", vector_a - vector_b)
# Dot product
print("Dot Product:", np.dot(vector_a, vector_b))
```

```python
# Define two matrices
matrix_a = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 1]
])
matrix_b = np.array([
    [1, 2, 0],
    [3, 1, 4],
    [2, 5, 1]
])
print("Matrix A:\n", matrix_a)
print("\nMatrix B:\n", matrix_b)
# Matrix addition
print("\nMatrix Addition:\n", matrix_a + matrix_b)
# Matrix multiplication
print("\nMatrix Multiplication:\n", np.dot(matrix_a, matrix_b))
```

```python
# Calculate determinant of Matrix A
determinant = np.linalg.det(matrix_a)
print("Determinant of Matrix A:", determinant)
# Calculate inverse if determinant is not zero
if not np.isclose(determinant, 0):
    inverse_matrix = np.linalg.inv(matrix_a)
    print("\nInverse of Matrix A:\n", inverse_matrix)
else:
    print("\nInverse does not exist because determinant is zero.")
```

```python
# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)
print("Eigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
```

### Interpretation

- Vector addition and subtraction perform element-wise mathematical operations.
- The dot product calculates the sum of the products of corresponding vector elements.
- Matrix addition adds corresponding elements, while matrix multiplication combines rows and columns.
- The determinant helps determine whether a matrix has an inverse.
- Eigenvalues and eigenvectors describe important properties of a matrix and are useful in data science techniques such as dimensionality reduction.

## Task 2 – Probability Concepts

In this task, we calculate sales probabilities, conditional probability, Bayes' theorem, and simulate probability distributions.

```python
# Probability of successful sales
sales_success = [0.65, 0.35]
success_probability = sales_success[0]
failure_probability = sales_success[1]
print("Probability of successful sales:", success_probability)
print("Probability of unsuccessful sales:", failure_probability)
```

```python
# Customer segment purchase probabilities
customer_segments = ["Premium", "Regular"]
segment_purchase_probability = {
    "Premium": 0.80,
    "Regular": 0.50
}
print("Customer Segments:", customer_segments)
for segment in customer_segments:
    print(
        f"Probability of purchase for {segment} customers:",
        segment_purchase_probability[segment]
    )
# Calculate overall purchase probability
premium_probability = 0.65
regular_probability = 0.35
overall_purchase_probability = (
    premium_probability * segment_purchase_probability["Premium"]
    + regular_probability * segment_purchase_probability["Regular"]
)
print("\nOverall probability of purchase:", overall_purchase_probability)
# Conditional probability: Premium customer given that a purchase occurred
premium_given_purchase = (
    premium_probability * segment_purchase_probability["Premium"]
) / overall_purchase_probability
print("Probability that a purchasing customer is Premium:",
      premium_given_purchase)
```

```python
# Bayes' theorem
P_A = 0.40
P_B = 0.62
P_B_given_A = 0.80
P_A_given_B = (P_B_given_A * P_A) / P_B
print("Probability of a customer being Premium given a purchase:",
      P_A_given_B)
```

```python
# Binomial distribution
n_trials = 20
success_prob = 0.65
binomial_data = stats.binom.rvs(
    n=n_trials,
    p=success_prob,
    size=1000,
    random_state=42
)
print("First 20 simulated binomial results:")
print(binomial_data[:20])
print("Average number of successes:", np.mean(binomial_data))
# Plot binomial distribution
plt.figure(figsize=(8, 5))
plt.hist(binomial_data, bins=range(0, 22), edgecolor="black")
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()
```

```python
# Normal distribution
normal_mean = 50
normal_std = 10
normal_data = stats.norm.rvs(
    loc=normal_mean,
    scale=normal_std,
    size=1000,
    random_state=42
)
print("Normal distribution sample:")
print(normal_data[:10])
plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=30, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
# Uniform distribution
uniform_low = 0
uniform_high = 1

uniform_data = stats.uniform.rvs(
    loc=uniform_low,
    scale=uniform_high - uniform_low,
    size=1000,
    random_state=42
)

print("Uniform distribution sample:")
print(uniform_data[:10])

plt.figure(figsize=(8, 5))
plt.hist(uniform_data, bins=30, edgecolor="black")
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation

- The probability of successful sales is 0.65, or 65%.
- Conditional probability helps us understand purchase behaviour within different customer segments.
- Bayes' theorem updates the probability of an event when new information is available.
- The binomial distribution represents the number of successes in a fixed number of trials.
- The normal distribution produces values around its mean.
- The uniform distribution gives equal probability across its specified interval.

## Task 3 – Descriptive Statistics

In this task, we analyse sales data using measures of central tendency, dispersion, skewness, and kurtosis.

```python
# Sales dataset
sales_data = [
    120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
    190, 200, 210, 195, 185, 175, 165, 155, 145, 135
]
# Convert the data into a NumPy array
sales_array = np.array(sales_data)
print("Sales Data:")
print(sales_array)
```

```python
from statistics import multimode
# Measures of central tendency
mean_value = np.mean(sales_array)
median_value = np.median(sales_array)
mode_value = multimode(sales_data)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

```python
# Measures of central tendency
mean_value = np.mean(sales_array)
median_value = np.median(sales_array)
mode_value = stats.mode(sales_array, keepdims=True)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value.mode)
```

```python
# Measures of dispersion
data_range = np.max(sales_array) - np.min(sales_array)
variance_value = np.var(sales_array)
standard_deviation = np.std(sales_array)
print("Minimum Value:", np.min(sales_array))
print("Maximum Value:", np.max(sales_array))
print("Range:", data_range)
print("Variance (Population):", variance_value)
print("Standard Deviation (Population):", standard_deviation)
```

```python
# Skewness and kurtosis
skewness_value = stats.skew(sales_array, bias=False)
kurtosis_value = stats.kurtosis(sales_array, bias=False)
print("Skewness:", skewness_value)
print("Kurtosis (Excess):", kurtosis_value)
```

```python
# Histogram of sales data
plt.figure(figsize=(8, 5))
plt.hist(sales_array, bins=8, edgecolor="black")
plt.title("Sales Data Distribution")
plt.xlabel("Sales Values")
plt.ylabel("Frequency")
plt.show()
# Box plot of sales data
plt.figure(figsize=(8, 5))
sns.boxplot(x=sales_array)
plt.title("Sales Data Box Plot")
plt.xlabel("Sales Values")
plt.show()
```

### Interpretation

- Mean represents the average sales value.
- Median represents the middle value when the data is arranged in order.
- Mode represents the most frequently occurring sales value. This dataset has multiple modes.
- Range measures the difference between the maximum and minimum sales values.
- Variance and standard deviation measure how spread out the sales values are.
- Skewness indicates whether the distribution is symmetrical or skewed.
- Kurtosis measures the shape and tail characteristics of the distribution.

The skewness value helps identify the direction of the distribution:
- Positive skewness indicates a right-skewed distribution.
- Negative skewness indicates a left-skewed distribution.
- Skewness close to zero indicates a relatively symmetrical distribution.
